```python
import random
```

```python
def newRoll():
  a = random.randrange(1,6)
  b = random.randrange(1,6)
  print(f"The sum of dice is {a} + {b} = {a + b}")
  return a,b
```

```python
a,b = newRoll()
if a + b == 7 or a + b == 11:
  print("You won")
elif a + b == 2 or a + b == 3 or a + b == 12:
  print(f"You lost")
else:
  g = a + b
  print(f"Your goal number is {g}")
  t = True
  while t:
    a,b = newRoll()
    if a + b == 7:
      print("You lost")
      t = False
    elif a + b == g:
      print("You won")
      t = False
```
